## What is RAG?
##### Retrieval-Augmented Generation (RAG) lets an LLM answer from your own data instead of relying only on what it learned in training. Instead of asking the model directly, you first retrieve relevant text from your documents and put it in the prompt as context. This reduces hallucination, keeps answers current without retraining, and lets you cite sources.

##### The pipeline has two phases:

- Indexing (offline): load documents → split into chunks → embed each chunk → store the vectors in a vector database.
- Query (online): embed the user's question → similarity search for the top-k chunks → pass question and chunks to the LLM → generate a grounded answer.

##### LangChain document loaders

#### Interface
#### Each loader can define its own parameters, but they all share two methods:

- load(): loads all documents at once.
- lazy_load(): streams documents one at a time, which helps with large datasets.

```python
from langchain_docling.loader import DoclingLoader
FILE_PATH = "content/attention_is_all_you_need.pdf"
loader = DoclingLoader(file_path = FILE_PATH)
```

```python
# Load all documents
documents = loader.load()
```
